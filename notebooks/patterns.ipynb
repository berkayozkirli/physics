```python
from physics.solvers.runge_kutta import RungeKutta4
from physics.solvers.spatial_discretization import laplacian2d_periodic, laplacian2d_reflective
from physics.systems.brusselator import Brusselator
from physics.systems.gray_scott import GrayScott
from physics.systems.schnakenberg import Schnakenberg
from physics.visualization.video import Video

import numpy as np
import numpy.typing as npt
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

## Gray - Scott

### Cheetah

```python
# Parameters
N_grid = 256
F, k, Du, Dv = 0.028, 0.062, 0.16, 0.08
t0, t_max, dt = 0.0, 50000.0, 1
name = "cheetah_spots_gs"
playback_speed = 500
fps = 60
colors='YlOrBr'
```

```python
print("Initializing Grid Simulation...")
u_steady,v_steady = 1.0, 0.0

center = N_grid // 2
seed_size = 10
u_initial = np.full((N_grid, N_grid), u_steady)
v_initial = np.full((N_grid, N_grid), v_steady)
u_initial[center-seed_size:center+seed_size, center-seed_size:center+seed_size] = 0.50
v_initial[center-seed_size:center+seed_size, center-seed_size:center+seed_size] = 0.25
noise = np.random.uniform(-0.01, 0.01, (seed_size*2, seed_size*2))
u_initial[center-seed_size:center+seed_size, center-seed_size:center+seed_size] += noise
v_initial[center-seed_size:center+seed_size, center-seed_size:center+seed_size] += noise
u_initial = np.clip(u_initial, 0.0, 1.0)
v_initial = np.clip(v_initial, 0.0, 1.0)

pde_system = GrayScott(F, k, laplacian2d_reflective, Du, Dv, dx=1.0)
solver = RungeKutta4(pde_system)
solver.dt = dt

video = Video(name=name, t0=t0, t_max=t_max, dt=dt, playback_speed=playback_speed, fps=fps)
total_steps = int((t_max - t0) / dt)
u_history = np.zeros((video.num_saves, N_grid, N_grid))

current_state = np.stack((u_initial, v_initial), axis=-1)
current_t = t0
save_idx = 0
print(f"Solving PDE with explicit loop: [{t0}, {t_max}] at {total_steps} steps...")
for step in range(total_steps):
    if step % video.save_interval == 0 and save_idx < video.num_saves:
        u_history[save_idx] = current_state[..., 0]
        save_idx += 1        
    current_state = solver.advance(current_t, current_state)
    current_t += dt
print("Simulation complete. Generating the video...")
video.create_from_grid(states=u_history, cmap=colors)
```

## Schnakenberg - Giraffe Patches

```python
# Parameters
N_grid = 256
a, b, Du, Dv = 0.1, 0.9, 1.0, 10.0
t0, t_max, dt = 0.0, 200.0, 0.01
name = "giraffe_s"
playback_speed = 2
fps = 60
colors = "copper"
```

```python
print("Initializing Grid Simulation...")
u_steady,v_steady = a+b, b / (a+b)**2

noise_scale = 0.01
u_noise = np.random.uniform(-noise_scale, noise_scale, (N_grid, N_grid))
v_noise = np.random.uniform(-noise_scale, noise_scale, (N_grid, N_grid))
u_initial = np.full((N_grid, N_grid), u_steady) + u_noise
v_initial = np.full((N_grid, N_grid), v_steady) + v_noise
u_initial = np.clip(u_initial, 0.0, None)
v_initial = np.clip(v_initial, 0.0, None)

pde_system = Schnakenberg(a, b, laplacian2d_reflective, Du, Dv, dx=1.0)
solver = RungeKutta4(pde_system)
solver.dt = dt

video = Video(name=name, t0=t0, t_max=t_max, dt=dt, playback_speed=playback_speed, fps=fps)
total_steps = int((t_max - t0) / dt)
u_history = np.zeros((video.num_saves, N_grid, N_grid))

current_state = np.stack((u_initial, v_initial), axis=-1)
current_t = t0
save_idx = 0
print(f"Solving PDE with explicit loop: [{t0}, {t_max}] at {total_steps} steps...")
for step in range(total_steps):
    if step % video.save_interval == 0 and save_idx < video.num_saves:
        u_history[save_idx] = current_state[..., 0]
        save_idx += 1        
    current_state = solver.advance(current_t, current_state)
    current_t += dt
print("Simulation complete. Generating the video...")
video.create_from_grid(states=u_history, cmap=colors)
```
